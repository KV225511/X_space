# AstroFoundation — Notebook 05: Multi-Modal Fusion Ablation Studies
### Systematic Evaluation of the 11 System Configurations (A1 – A11)

Demonstrating the empirical value of:
1. Multi-modal information synergy over unimodal baselines.
2. Pairwise Cross-Attention vs naive feature concatenation.
3. Learned $\sigma$-Gating network.
4. Stochastic Modality Dropout ($p=0.15$).
5. Monte Carlo Dropout Bayesian Uncertainty calibration.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_ablations = pd.read_csv('../outputs/ablation_results.csv')
df_ablations
```

## 1. Macro F1 Across Ablation Configurations

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=df_ablations, x='Config ID', y='Macro F1', palette='viridis')
plt.title('AstroFoundation — Ablation Study: Macro F1 Across Configurations', fontsize=13)
plt.ylabel('Macro F1 Score')
plt.xlabel('Ablation ID (A1..A11)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## 2. Expected Calibration Error (ECE) Comparison

```python
plt.figure(figsize=(10, 4))
sns.lineplot(data=df_ablations, x='Config ID', y='ECE (%)', marker='o', color='crimson')
plt.title('Expected Calibration Error (%) Across Model Variations', fontsize=13)
plt.ylabel('ECE (%) [Lower is Better]')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
