## Dummy Spectrum Exploration

In this section, a simple artificial spectrum was created to understand the basic structure of astronomical spectral data before working with a real SDSS spectrum.

A spectrum describes how the measured brightness of an astronomical object changes with wavelength.

Conceptually:

* The **x-axis** represents wavelength.
* The **y-axis** represents flux, or the measured brightness at each wavelength.

### Wavelength Array

The wavelength values were generated from `4000 Å` to `7000 Å` using 1000 evenly spaced samples.

This range covers a large part of the visible-light region.

The resulting wavelength array has shape:

```text
(1000,)
```

This means that the spectrum contains 1000 wavelength positions.

The spacing between adjacent wavelength values is approximately:

```text
3.003 Å
```

Astropy units were attached using `u.AA`, so the wavelength array became an `astropy.units.Quantity` rather than a plain NumPy array.

This is important because scientific code should preserve physical units and prevent accidental mixing of incompatible quantities.

### Flat Flux Baseline

A flat flux array was initially created using `np.ones()`.

Every flux value was set to:

```text
1 Jy
```

where `Jy` stands for Jansky, a unit of spectral flux density.

The flux array also has shape:

```text
(1000,)
```

The wavelength and flux arrays must have the same length because every wavelength value needs one corresponding flux measurement.

The relationship is:

```text
wavelength[i] ↔ flux[i]
```

A flat flux of `1 Jy` represents an idealized spectrum with the same brightness at every wavelength.

This is not a realistic stellar spectrum, but it provides a simple baseline for understanding how spectral features modify the measured flux.

### Artificial Absorption Line

A Gaussian-shaped absorption line was added around:

```text
5500 Å
```

The absorption profile was calculated using:

```text
absorption = depth × exp(-((wavelength - center)²) / (2 × width²))
```

The chosen parameters were:

```text
center = 5500 Å
width = 40 Å
depth = 0.4
```

The meaning of each parameter is:

* `center` determines the wavelength where the absorption line is strongest.
* `width` controls how broad or narrow the line is.
* `depth` controls how much flux is removed at the center.

The original flat flux was `1 Jy`.

The maximum absorption was approximately:

```text
0.3997
```

Therefore, the minimum modified flux became approximately:

```text
1.0 - 0.3997 = 0.6003 Jy
```

The minimum absorption value was extremely close to zero at wavelengths far away from `5500 Å`.

This means the absorption feature affects only a limited wavelength region, while the rest of the spectrum remains close to the original `1 Jy` baseline.

### Scientific Interpretation

An absorption line appears when atoms or molecules between the light source and the observer absorb photons at specific wavelengths.

In a real stellar spectrum, absorption lines can provide information about:

* Chemical composition
* Temperature
* Pressure
* Radial velocity
* Stellar rotation
* Motion of gas in the star's atmosphere

The artificial line created here is a simplified Gaussian approximation.

Real spectral lines may have more complex shapes because of instrumental effects, Doppler broadening, pressure broadening, noise, and overlapping atomic transitions.

### Technical Interpretation

The final dummy spectrum contains:

```text
1000 wavelength measurements
1000 corresponding flux measurements
wavelength unit: Angstrom
flux unit: Jansky
one artificial Gaussian absorption line
```

The wavelength and flux arrays are both one-dimensional and aligned by index.

This makes them suitable for storing inside a `Spectrum1D` object from `specutils`.

The dummy spectrum helps verify the following before working with real data:

1. Wavelength and flux arrays must have matching shapes.
2. Physical units should be preserved.
3. Spectral lines appear as peaks or dips in flux.
4. Gaussian functions can simulate simplified spectral features.
5. The spectrum can be plotted as wavelength versus flux.

This controlled example provides a clear reference for identifying similar structures in a real SDSS spectrum.

````python
flux = np.asarray(flux, dtype=np.float32)
loglam = np.asarray(loglam, dtype=np.float32)
ivar = np.asarray(ivar, dtype=np.float32)
````

````python
# ============================================================
# 1. Extract the COADD data
# ============================================================

coadd_data = spectrum_hdul[1].data

flux = np.asarray(coadd_data["flux"], dtype=np.float32)
loglam = np.asarray(coadd_data["loglam"], dtype=np.float32)
ivar = np.asarray(coadd_data["ivar"], dtype=np.float32)
model_flux = np.asarray(coadd_data["model"], dtype=np.float32)
sky_flux = np.asarray(coadd_data["sky"], dtype=np.float32)

print("Flux shape:", flux.shape)
print("Loglam shape:", loglam.shape)
print("Inverse variance shape:", ivar.shape)
print("Flux dtype:", flux.dtype)
````

````python
# ============================================================
# 2. Convert logarithmic wavelength into Angstrom
# ============================================================

wavelength = 10 ** loglam

print("Wavelength shape:", wavelength.shape)
print("First 5 wavelengths:", wavelength[:5])
print("Minimum wavelength:", wavelength.min())
print("Maximum wavelength:", wavelength.max())
print("First wavelength spacing:", wavelength[1] - wavelength[0])
print("Last wavelength spacing:", wavelength[-1] - wavelength[-2])
````

````python
# ============================================================
# 3. Check for invalid values
# ============================================================

print("NaNs in wavelength:", np.isnan(wavelength).sum())
print("NaNs in flux:", np.isnan(flux).sum())
print("NaNs in ivar:", np.isnan(ivar).sum())

print("Infinite wavelength values:", np.isinf(wavelength).sum())
print("Infinite flux values:", np.isinf(flux).sum())
print("Infinite ivar values:", np.isinf(ivar).sum())

print("Points with ivar <= 0:", np.sum(ivar <= 0))
````

````python
# ============================================================
# 4. Create a mask for valid spectral points
# ============================================================

valid_mask = (
    np.isfinite(wavelength)
    & np.isfinite(flux)
    & np.isfinite(ivar)
    & (ivar > 0)
)

clean_wavelength = wavelength[valid_mask]
clean_flux = flux[valid_mask]
clean_ivar = ivar[valid_mask]
clean_model_flux = model_flux[valid_mask]
clean_sky_flux = sky_flux[valid_mask]

print("Original points:", len(wavelength))
print("Valid points:", len(clean_wavelength))
print("Removed points:", len(wavelength) - len(clean_wavelength))
````

````python
# ============================================================
# 5. Calculate uncertainty from inverse variance
# ============================================================

flux_error = 1 / np.sqrt(clean_ivar)

print("First 5 inverse variance values:", clean_ivar[:5])
print("First 5 estimated errors:", flux_error[:5])
````

````python
# ============================================================
# 6. Inspect the clean spectrum
# ============================================================

print("Clean wavelength range:")
print(clean_wavelength.min(), "to", clean_wavelength.max(), "Angstrom")

print("\nClean flux range:")
print(clean_flux.min(), "to", clean_flux.max())

print("\nMean flux:", np.mean(clean_flux))
print("Median flux:", np.median(clean_flux))
print("Standard deviation:", np.std(clean_flux))
````

````python
# ============================================================
# 7. Plot the raw real SDSS spectrum
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    clean_wavelength,
    clean_flux,
    linewidth=0.8,
    label="Observed flux"
)

plt.xlabel("Wavelength (Angstrom)")
plt.ylabel(r"Flux ($10^{-17}$ erg s$^{-1}$ cm$^{-2}$ Angstrom$^{-1}$)")
plt.title("Real SDSS Spectrum")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
````

````python
# ============================================================
# 8. Compare observed flux with the SDSS model
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    clean_wavelength,
    clean_flux,
    linewidth=0.7,
    label="Observed flux"
)

plt.plot(
    clean_wavelength,
    clean_model_flux,
    linewidth=1.0,
    label="SDSS fitted model"
)

plt.xlabel("Wavelength (Angstrom)")
plt.ylabel(r"Flux ($10^{-17}$ erg s$^{-1}$ cm$^{-2}$ Angstrom$^{-1}$)")
plt.title("Observed SDSS Spectrum vs Pipeline Model")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
````

````python
# ============================================================
# 9. Create a specutils Spectrum object
# ============================================================

from specutils import Spectrum
import astropy.units as u
from astropy.nddata import StdDevUncertainty

flux_unit = u.Unit("1e-17 erg / (s cm2 Angstrom)")

real_spectrum = Spectrum(
    spectral_axis=clean_wavelength * u.AA,
    flux=clean_flux * flux_unit,
    uncertainty=StdDevUncertainty(flux_error * flux_unit)
)

print(real_spectrum)
print("Spectral axis shape:", real_spectrum.spectral_axis.shape)
print("Flux shape:", real_spectrum.flux.shape)
print("Wavelength unit:", real_spectrum.spectral_axis.unit)
print("Flux unit:", real_spectrum.flux.unit)
````

````python
# ============================================================
# 10. Estimate the continuum using a median filter
# ============================================================

from scipy.ndimage import median_filter

continuum = median_filter(
    clean_flux,
    size=101,
    mode="nearest"
)

print("Continuum shape:", continuum.shape)
print("Continuum minimum:", continuum.min())
print("Continuum maximum:", continuum.max())
````

````python
# ============================================================
# 11. Prevent division by zero or extremely small values
# ============================================================

continuum_safe = continuum.copy()

minimum_safe_value = 1e-6
continuum_safe[np.abs(continuum_safe) < minimum_safe_value] = minimum_safe_value
````

````python
# ============================================================
# 12. Normalize the spectrum
# ============================================================

normalized_flux = clean_flux / continuum_safe

print("Normalized shape:", normalized_flux.shape)
print("Normalized minimum:", normalized_flux.min())
print("Normalized maximum:", normalized_flux.max())
print("Normalized median:", np.median(normalized_flux))
````

````python
# ============================================================
# 13. Plot the estimated continuum
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    clean_wavelength,
    clean_flux,
    linewidth=0.7,
    label="Observed flux"
)

plt.plot(
    clean_wavelength,
    continuum,
    linewidth=1.5,
    label="Estimated continuum"
)

plt.xlabel("Wavelength (Angstrom)")
plt.ylabel(r"Flux ($10^{-17}$ erg s$^{-1}$ cm$^{-2}$ Angstrom$^{-1}$)")
plt.title("SDSS Spectrum with Estimated Continuum")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
````

````python
# ============================================================
# 14. Plot the continuum-normalized spectrum
# ============================================================

plt.figure(figsize=(12, 5))

plt.plot(
    clean_wavelength,
    normalized_flux,
    linewidth=0.8
)

plt.axhline(
    y=1.0,
    linestyle="--",
    linewidth=1,
    label="Normalized continuum"
)

plt.xlabel("Wavelength (Angstrom)")
plt.ylabel("Normalized Flux")
plt.title("Continuum-Normalized SDSS Spectrum")
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
````

````python
# ============================================================
# 15. Create a normalized Spectrum object
# ============================================================

normalized_spectrum = Spectrum(
    spectral_axis=clean_wavelength * u.AA,
    flux=normalized_flux * u.dimensionless_unscaled
)

print(normalized_spectrum)
````

````python
# ============================================================
# 16. Final validation checks
# ============================================================

assert clean_wavelength.shape == clean_flux.shape
assert clean_flux.shape == clean_ivar.shape
assert normalized_flux.shape == clean_flux.shape

assert clean_wavelength.dtype == np.float32
assert clean_flux.dtype == np.float32
assert clean_ivar.dtype == np.float32

assert np.all(np.isfinite(clean_wavelength))
assert np.all(np.isfinite(clean_flux))
assert np.all(np.isfinite(clean_ivar))
assert np.all(clean_ivar > 0)
assert np.all(np.isfinite(normalized_flux))

print("All real-spectrum validation checks passed.")
````

## Real SDSS Spectrum Exploration

A real spectrum was downloaded from the Sloan Digital Sky Survey using `astroquery.sdss`.

The SDSS query first returned observation metadata. The actual spectrum was then downloaded using `SDSS.get_spectra()`, which returned a FITS `HDUList`.

The main combined spectrum was stored in the `COADD` extension. This table contained 3823 rows, with each row representing one wavelength measurement.

The important columns were:

* `flux`: measured brightness at each wavelength
* `loglam`: base-10 logarithm of wavelength
* `ivar`: inverse variance representing measurement reliability
* `sky`: estimated background sky contribution
* `model`: spectrum fitted by the SDSS processing pipeline

The actual wavelength was reconstructed using:

```text
wavelength = 10 ** loglam
```

The wavelength and flux arrays both had shape `(3823,)`, meaning every wavelength position had one corresponding flux measurement.

Invalid points were removed using the following conditions:

* wavelength must be finite;
* flux must be finite;
* inverse variance must be finite;
* inverse variance must be greater than zero.

The uncertainty of each valid flux measurement was estimated from inverse variance using:

```text
error = 1 / sqrt(ivar)
```

The cleaned wavelength, flux, and uncertainty arrays were placed inside a `specutils.Spectrum` object.

### Continuum Normalization

A real astronomical spectrum contains a broad underlying shape called the continuum. Spectral absorption and emission lines appear as local variations around this broad trend.

A median filter was used to estimate the continuum. The original flux was then divided by the estimated continuum:

```text
normalized flux = observed flux / continuum
```

After normalization, the continuum lies approximately around `1`.

* Absorption lines generally appear below `1`.
* Emission lines generally appear above `1`.

Continuum normalization makes spectra with different overall brightness levels easier to compare. It also allows a machine-learning model to focus more strongly on spectral-line structures rather than the absolute brightness scale.

The final cleaned and normalized spectrum is suitable for later resampling, spectral feature extraction, PCA reduction, and conversion into the fixed-size input expected by the spectral encoder.

# 1. AstroFoundation's Three Data Modalities

AstroFoundation uses three different kinds of astronomical observations:

1. **Light curves**
2. **Telescope pixel images**
3. **Spectra**

Although all three may be stored in FITS files, their scientific meaning and array structure are different.

## 1.1 Light Curve

A light curve measures how the brightness of an astronomical object changes with time.

```text
x-axis: time
y-axis: measured flux or brightness
```

A light curve can reveal:

- Exoplanet transits
- Variable stars
- Eclipsing binaries
- Stellar flares
- Periodic brightness changes

Typical raw relationship:

```text
time[i] ↔ flux[i]
```

The planned processed shape is:

```text
(1024,)
```

A batch entering the model may have shape:

```text
(batch_size, 1024)
```

Inside the Transformer, it can be converted to:

```text
(batch_size, 1024, 1)
```

The final `LightCurveEncoder` produces:

```text
(batch_size, 256)
```

---

## 1.2 Telescope Target-Pixel Image

A target-pixel file contains a sequence of small telescope images around a target star.

Each pixel stores the amount of light detected at one location on the telescope sensor.

A Kepler target-pixel flux array can have shape:

```text
(number_of_times, height, width)
```

Example:

```text
(1639, 8, 7)
```

This means:

- 1639 observation times
- 8 detector-pixel rows
- 7 detector-pixel columns

One image frame has shape:

```text
(8, 7)
```

After preprocessing, the image is converted into three channels and resized:

```text
(3, 224, 224)
```

A batch entering an image encoder has shape:

```text
(batch_size, 3, 224, 224)
```

The final image encoder produces:

```text
(batch_size, 256)
```

---

## 1.3 Spectrum

A spectrum measures how much light is detected at different wavelengths.

```text
x-axis: wavelength
y-axis: spectral flux
```

A spectrum can reveal:

- Chemical composition
- Temperature
- Redshift
- Radial velocity
- Gas motion
- Stellar rotation
- Absorption lines
- Emission lines

The basic relationship is:

```text
wavelength[i] ↔ flux[i]
```

A real SDSS spectrum in this notebook contained:

```text
3823 wavelength measurements
3823 flux measurements
3823 inverse-variance measurements
```

The planned spectral pipeline later reduces the cleaned spectrum to:

```text
(128,)
```

The `SpectralEncoder` then produces:

```text
(batch_size, 256)
```

# 2. FITS Files

## 2.1 Full Form

FITS stands for:

```text
Flexible Image Transport System
```

It is an astronomy-standard file format used to store:

- Images
- Spectra
- Tables
- Time-series observations
- Aperture masks
- Telescope metadata
- Coordinates
- Calibration information

A FITS file is a **container format**. It does not imply that the file contains only an image.

The same FITS format can store completely different scientific data.

---

## 2.2 HDU Full Form

HDU stands for:

```text
Header Data Unit
```

Every HDU contains:

```text
HDU = Header + Data
```

The header stores metadata.

The data section stores the actual image, table, mask, or other scientific values.

A FITS file contains an ordered list of HDUs:

```text
FITS file
├── HDU 0
├── HDU 1
├── HDU 2
└── ...
```

When Astropy opens a FITS file:

```python
hdul = fits.open(path)
```

`hdul` is an `HDUList`.

It behaves similarly to a Python list:

```python
hdul[0]
hdul[1]
hdul[2]
```

---

## 2.3 Common HDU Types

### PrimaryHDU

Usually stored at index 0.

```python
primary_hdu = hdul[0]
```

It always has a header but may not contain a data array.

Typical use:

- Object name
- Telescope name
- Instrument details
- Coordinates
- Observation metadata

### ImageHDU

Stores image-like NumPy arrays.

Example:

```python
aperture_mask = hdul[2].data
```

### BinTableHDU

Stores a binary table with rows and named columns.

Example:

```python
table = hdul[1].data
flux = table["FLUX"]
```

---

## 2.4 Header and Data

Every HDU can be inspected using:

```python
hdu.header
hdu.data
```

The header works similarly to a dictionary:

```python
header["OBJECT"]
header["RA_OBJ"]
header["DEC_OBJ"]
```

The `.data` structure depends on the HDU type:

- `ImageHDU.data` may be a 2D image.
- `BinTableHDU.data` may be a structured table.
- `PrimaryHDU.data` may be `None`.

# 4. FITS Axis-Order Difference

The FITS output displayed:

```text
Dimensions: (7, 8)
```

but NumPy displayed:

```text
shape: (8, 7)
```

This is not an error.

FITS commonly reports axes as:

```text
(width, height)
```

NumPy reports a 2D array as:

```text
(height, width)
```

Therefore:

```text
FITS:  (7, 8)
NumPy: (8, 7)
```

refer to the same image.

# 5. Endianness

The aperture and flux arrays had dtypes such as:

```text
>i4
>f4
```

## 5.1 Meaning of `>i4`

```text
>  = big-endian
i  = signed integer
4  = 4 bytes
```

Therefore:

```text
>i4 = big-endian 32-bit signed integer
```

## 5.2 Meaning of `>f4`

```text
>  = big-endian
f  = floating point
4  = 4 bytes
```

Therefore:

```text
>f4 = big-endian 32-bit floating point
```

FITS traditionally stores values in big-endian byte order.

Most modern computers use native little-endian byte order.

A convenient conversion is:

```python
array = np.asarray(array, dtype=np.float32)
```

This creates a native-endian `float32` array suitable for PyTorch.

# 6. FITS Image Preprocessing

The image preprocessing pipeline was:

```text
raw flux frame
↓
convert to native float32
↓
replace NaN and infinite values
↓
percentile clipping
↓
min-max normalization
↓
three-channel conversion
↓
resize to 224 × 224
```

---

## 6.1 Handling NaN Values

NaN means:

```text
Not a Number
```

NaNs cannot safely be passed into a neural network because they can propagate through calculations and make the output or loss become NaN.

A median fill can be used:

```python
fill_value = np.nanmedian(frame)

clean_frame = np.nan_to_num(
    frame,
    nan=fill_value,
    posinf=fill_value,
    neginf=fill_value
)
```

---

## 6.2 Percentile Clipping

The frame may have an extreme brightness range.

A few very bright pixels can dominate the entire normalization.

Percentile clipping limits extreme values.

```python
low = np.percentile(clean_frame, 1)
high = np.percentile(clean_frame, 99)

clipped_frame = np.clip(clean_frame, low, high)
```

This retains the central 1st-to-99th-percentile range.

Scientific purpose:

- Prevent a few extreme pixels from dominating the image display.
- Preserve useful star-region structure.

Technical purpose:

- Stabilize normalization.
- Reduce the effect of outliers.
- Produce better-behaved neural-network input.

---

## 6.3 Min-Max Normalization

Formula:

```text
x_normalized = (x - x_min) / (x_max - x_min)
```

A numerically safer implementation adds a small value:

```python
normalized_frame = (
    clipped_frame - clipped_frame.min()
) / (
    clipped_frame.max() - clipped_frame.min() + 1e-8
)
```

The result should lie between:

```text
0 and 1
```

# 7. `np.stack()`

## 7.1 Purpose

`np.stack()` combines arrays by creating a **new axis**.

Syntax:

```python
np.stack(arrays, axis=0)
```

The first argument is usually a Python list or tuple containing arrays:

```python
np.stack([array1, array2, array3], axis=0)
```

All arrays must have the same shape.

---

## 7.2 Three-Channel Conversion

The original flux frame has shape:

```text
(8, 7)
```

It is a single-channel grayscale image.

Pretrained ResNet and ViT models generally expect:

```text
(3, height, width)
```

The image is duplicated three times:

```python
flux_3ch = np.stack(
    [normalized_frame] * 3,
    axis=0
)
```

The expression:

```python
[normalized_frame] * 3
```

creates:

```text
[
    normalized_frame,
    normalized_frame,
    normalized_frame
]
```

`axis=0` inserts the new channel axis at the beginning.

Result:

```text
(3, 8, 7)
```

Interpretation:

```text
3 channels
8 rows
7 columns
```

The channels are identical. This does not add new scientific information; it only makes the grayscale data compatible with pretrained three-channel models.

---

## 7.3 Stack versus Concatenate

`stack` creates a new axis.

```python
np.stack([a, b], axis=0)
```

If each input has shape `(3,)`, the result has shape:

```text
(2, 3)
```

`concatenate` joins along an existing axis.

```python
np.concatenate([a, b], axis=0)
```

If each input has shape `(3,)`, the result has shape:

```text
(6,)
```

Summary:

```text
stack       → creates a new dimension
concatenate → extends an existing dimension
```

# 11. Gaussian Absorption Line

A synthetic absorption line was created using:

```text
absorption = depth × exp(
    -((wavelength - center)^2) / (2 × width^2)
)
```

Python implementation:

```python
absorption = depth * np.exp(
    -((wavelength.value - center) ** 2)
    / (2 * width ** 2)
)
```

Parameters:

```text
center = 5500 Å
width  = 40 Å
depth  = 0.4
```

---

## 11.1 Center

The center determines where the line is strongest.

```text
center = 5500 Å
```

At or near this wavelength, maximum absorption occurs.

---

## 11.2 Depth

Depth controls how much flux is removed at the line center.

With:

```text
baseline flux = 1.0 Jy
depth = 0.4
```

the central flux becomes approximately:

```text
1.0 - 0.4 = 0.6 Jy
```

---

## 11.3 Width

Width controls how narrow or broad the Gaussian line is.

Technically, it acts like the Gaussian standard deviation, commonly denoted by:

```text
σ
```

- Small width → narrow, sharp line
- Large width → broad line

Width is not exactly the complete visible width.

The Gaussian full width at half maximum is:

```text
FWHM = 2 × sqrt(2 × ln(2)) × σ
```

Approximately:

```text
FWHM ≈ 2.355 × σ
```

For `width = 40 Å`:

```text
FWHM ≈ 94.2 Å
```

Scientifically, spectral-line broadening can be influenced by:

- Temperature
- Gas pressure
- Stellar rotation
- Radial motions
- Instrument resolution

---

## 11.4 `np.exp()`

Syntax:

```python
np.exp(x)
```

This calculates:

```text
e^x
```

where:

```text
e ≈ 2.71828
```

Examples:

```text
exp(0) = 1
exp(1) ≈ 2.71828
```

For an array, it is applied element by element.

In the Gaussian formula:

- At the center, `wavelength - center = 0`.
- Therefore, the exponent is 0.
- `exp(0) = 1`.
- Absorption becomes approximately equal to `depth`.

Far from the center:

- The exponent becomes a large negative value.
- `exp(-large value)` becomes almost zero.
- Therefore, the absorption disappears.

# 15. COADD Columns

## 15.1 Observed Flux

```python
flux = coadd_data["flux"]
```

This is the calibrated observed spectrum attributed to the astronomical target.

It contains:

- Real object signal
- Absorption lines
- Emission lines
- Measurement noise
- Possible remaining calibration errors

Important:

```text
Observed flux is not the same as normalized flux.
```

---

## 15.2 `loglam`

SDSS stores wavelength as:

```text
loglam = log10(wavelength in Å)
```

Recover wavelength using:

```text
wavelength = 10^loglam
```

Python:

```python
wavelength = 10 ** loglam
```

Why use logarithmic wavelength spacing?

Equal steps in logarithmic wavelength correspond approximately to equal velocity intervals.

For small Doppler shifts:

```text
Δv ≈ c × Δλ / λ
```

Since:

```text
Δ(log λ) ≈ Δλ / (λ ln 10)
```

uniform log-wavelength sampling is useful for:

- Doppler-shift analysis
- Radial-velocity calculations
- Template matching
- Comparing spectra across wavelengths

---

## 15.3 Inverse Variance

```python
ivar = coadd_data["ivar"]
```

Variance is:

```text
variance = σ^2
```

Inverse variance is:

```text
ivar = 1 / σ^2
```

Therefore, uncertainty can be recovered using:

```text
σ = 1 / sqrt(ivar)
```

Python:

```python
error = 1 / np.sqrt(ivar)
```

Interpretation:

- Large `ivar` → smaller uncertainty → more reliable measurement
- Small `ivar` → larger uncertainty → less reliable measurement
- `ivar = 0` → invalid or unusable measurement

Only calculate the error where:

```text
ivar > 0
```

---

## 15.4 Sky Flux

```python
sky_flux = coadd_data["sky"]
```

The telescope receives both target light and background sky light.

Conceptually:

```text
detected signal =
target light + sky background + noise
```

Possible sky-background sources:

- Atmospheric airglow
- Moonlight
- Artificial light
- Night-sky emission lines
- Scattered light

The SDSS pipeline estimates the sky contribution.

The published object flux is generally already sky-subtracted, so the `sky` column should normally not be subtracted again.

It is useful for identifying wavelengths where strong sky features may affect the observation.

---

## 15.5 Model Flux

```python
model_flux = coadd_data["model"]
```

This is not another telescope measurement.

It is the best-fitting spectrum produced by the SDSS analysis pipeline.

Conceptually:

```text
observed flux
↓
template/model fitting
↓
model flux
```

The residual is:

```text
residual = observed flux - model flux
```

Python:

```python
residual = flux - model_flux
```

A large residual can indicate:

- Noise
- Poor model fit
- Unusual spectral feature
- Calibration issue
- Template mismatch

---

## 15.6 Quality Masks

`and_mask` indicates issues present in all contributing exposures.

`or_mask` indicates issues present in at least one contributing exposure.

They are bitmasks, so individual bits encode different quality conditions.

A production pipeline may use them to remove known bad spectral points.

---

## 15.7 Wavelength Dispersion

`wdisp` contains information related to spectral resolution or wavelength dispersion.

It describes how the instrument spreads light in wavelength space.

A larger instrumental spread can make a spectral line appear broader.

# 17. Continuum and Continuum Normalization

## 17.1 What Is the Continuum?

The continuum is the broad, slowly changing baseline shape of a spectrum.

Spectral lines appear as local variations around it.

```text
continuum
├── absorption line dips below it
└── emission line rises above it
```

The continuum can be influenced by:

- Object temperature
- Overall energy distribution
- Dust extinction
- Telescope response
- Calibration
- Redshift

---

## 17.2 Continuum Estimation

A simple median filter was used:

```python
continuum = median_filter(
    clean_flux,
    size=101,
    mode="nearest"
)
```

The filter estimates the local median across a broad window.

Why a broad window?

- Narrow spectral lines should not define the continuum.
- The filter should follow the broad baseline instead.

The value `101` is a practical demonstration choice, not a universally optimal scientific setting.

A more advanced pipeline might use:

- Polynomial fitting
- Splines
- Iterative sigma clipping
- Specutils continuum fitting
- Domain-specific masks for known lines

---

## 17.3 Continuum-Normalized Flux

Formula:

```text
normalized flux =
observed flux / estimated continuum
```

Python:

```python
normalized_flux = clean_flux / continuum
```

After normalization:

```text
continuum ≈ 1
absorption lines < 1
emission lines > 1
```

Scientific purpose:

- Compare line shapes across objects with different brightness scales.
- Focus on relative line features.
- Reduce broad spectral-shape differences.

Technical purpose:

- Produce a more consistent input scale.
- Help the model focus on local spectral patterns.
- Reduce sensitivity to absolute brightness.

# 18. Observed, Sky, Model, and Normalized Flux

The four terms are different:

```text
observed flux
    = calibrated light measured for the target

sky flux
    = estimated background light from the sky

model flux
    = SDSS pipeline's fitted expected spectrum

normalized flux
    = observed flux divided by estimated continuum
```

At each wavelength index:

```text
wavelength[i]
├── flux[i]            observed target value
├── sky[i]             background estimate
├── model[i]           fitted model value
└── normalized_flux[i] continuum-corrected observed value
```

The planned neural-network input should generally come from cleaned, consistently processed observed data rather than directly using the SDSS fitted model.

Using model flux as the target input could make AstroFoundation learn SDSS's existing model rather than learn directly from observations.

# 23. Why Every Encoder Outputs 256 Dimensions

Each modality begins with a different data structure:

```text

image       → 2D spatial data
spectrum    → 1D wavelength sequence/features
```

The encoders transform them into the same representation size:

```text

```

This common embedding size allows the multimodal fusion module to:

- Compare modalities
- Apply projection heads
- Apply cross-attention
- Learn modality weights
- Compute a fused representation

Without a shared dimension, cross-modal operations would require additional projections before the modalities could interact.

# 24. Important Scientific and Technical Cautions

1. A FITS aperture value of `3` indicates aperture selection, not the physical star boundary.

2. Percentile clipping is a preprocessing decision and must be tested rather than assumed optimal.

3. Repeating a grayscale frame into three channels adds compatibility, not new information.

4. `model` flux is an SDSS fitted result and should not automatically replace observed flux.

5. The SDSS `flux` column is generally already sky-subtracted; do not subtract `sky` again without verifying the data documentation.

6. A median-filter continuum is useful for EDA but may be too simple for the final scientific pipeline.

7. PCA must be fitted only on the training set to avoid data leakage.

8. All spectra must eventually be placed on a consistent wavelength grid before comparing or batching them.

9. Missing wavelengths and redshift must be handled carefully.

10. Assertions help validate assumptions, but proper tests and explicit error handling are also required in production code.
